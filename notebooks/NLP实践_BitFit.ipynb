```python
# ============================================================
# Cell 4
# 完整实验函数
# ============================================================

import os
import sys
import json
import time
import random
import copy
import traceback

import numpy as np
import torch

from pathlib import Path

# ------------------------------------------------------------
# 路径
# ------------------------------------------------------------

BITFIT_DIR = "/content/BitFit"

DRIVE_ROOT = "/content/drive/MyDrive/nlp实践"

MODEL_ROOT = os.path.join(
    DRIVE_ROOT,
    "BitFit_models"
)

RESULT_ROOT = os.path.join(
    DRIVE_ROOT,
    "BitFit_results"
)

RESULTS_PATH = os.path.join(
    RESULT_ROOT,
    "bitfit_stage_b.json"
)

os.makedirs(MODEL_ROOT, exist_ok=True)
os.makedirs(RESULT_ROOT, exist_ok=True)

os.chdir(BITFIT_DIR)

# ------------------------------------------------------------
# 全局结果
# ------------------------------------------------------------

all_results = []

# ------------------------------------------------------------
# 读取已经存在的结果
# ------------------------------------------------------------

if os.path.exists(RESULTS_PATH):

    try:

        with open(
            RESULTS_PATH,
            "r",
            encoding="utf-8"
        ) as f:

            all_results = json.load(f)

        print(
            "发现已有结果：",
            len(all_results),
            "个实验"
        )

    except Exception as e:

        print(
            "已有 JSON 读取失败，将重新开始结果列表：",
            repr(e)
        )

        all_results = []

else:

    print("没有发现旧结果，从零开始。")


# ------------------------------------------------------------
# 保存 JSON
# ------------------------------------------------------------

def save_results():

    tmp_path = RESULTS_PATH + ".tmp"

    with open(
        tmp_path,
        "w",
        encoding="utf-8"
    ) as f:

        json.dump(
            all_results,
            f,
            indent=2,
            ensure_ascii=False
        )

    os.replace(
        tmp_path,
        RESULTS_PATH
    )


# ------------------------------------------------------------
# 实验 key
# ------------------------------------------------------------

def experiment_key(
    method,
    lr,
    seed
):

    return (
        str(method),
        float(lr),
        int(seed)
    )


# ------------------------------------------------------------
# 判断是否已经完成
# ------------------------------------------------------------

def has_completed(
    method,
    lr,
    seed
):

    key = experiment_key(
        method,
        lr,
        seed
    )

    for r in all_results:

        try:

            old_key = experiment_key(
                r["method"],
                r["lr"],
                r["seed"]
            )

            if old_key == key:
                return True

        except Exception:
            pass

    return False


# ------------------------------------------------------------
# 参数快照
# ------------------------------------------------------------

def snapshot_trainable_parameters(model):

    result = {}

    for name, param in model.named_parameters():

        if param.requires_grad:

            result[name] = (
                param.detach()
                .cpu()
                .clone()
            )

    return result


# ------------------------------------------------------------
# 检查参数是否发生变化
# ------------------------------------------------------------

def check_parameter_updates(
    model,
    before
):

    updated = []
    not_updated = []

    max_change = 0.0

    for name, param in model.named_parameters():

        if not param.requires_grad:
            continue

        after = (
            param.detach()
            .cpu()
        )

        if name not in before:

            continue

        old = before[name]

        diff = (
            after - old
        ).abs()

        change = diff.max().item()

        max_change = max(
            max_change,
            change
        )

        if torch.any(diff != 0):

            updated.append(name)

        else:

            not_updated.append(name)

    return (
        updated,
        not_updated,
        max_change
    )


# ------------------------------------------------------------
# 检查梯度
# ------------------------------------------------------------

def check_gradients(model):

    with_grad = []
    without_grad = []

    max_grad = 0.0

    for name, param in model.named_parameters():

        if not param.requires_grad:
            continue

        if param.grad is None:

            without_grad.append(name)

        else:

            with_grad.append(name)

            value = (
                param.grad.detach()
                .abs()
                .max()
                .item()
            )

            max_grad = max(
                max_grad,
                value
            )

    return (
        with_grad,
        without_grad,
        max_grad
    )


# ------------------------------------------------------------
# optimizer steps
# ------------------------------------------------------------

def get_optimizer_steps(optimizer):

    steps = []

    for state in optimizer.state.values():

        if "step" not in state:
            continue

        step = state["step"]

        if torch.is_tensor(step):
            step = step.item()

        steps.append(float(step))

    return sorted(
        set(steps)
    )


# ------------------------------------------------------------
# 保存模型
# ------------------------------------------------------------

def save_model_to_drive(
    ev,
    method,
    lr,
    seed
):

    folder_name = (
        method
        .replace(" ", "_")
        .replace("/", "_")
    )

    folder_name += (
        f"_lr{lr:g}_seed{seed}"
    )

    model_dir = os.path.join(
        MODEL_ROOT,
        folder_name
    )

    os.makedirs(
        model_dir,
        exist_ok=True
    )

    ev.model.save_pretrained(
        model_dir
    )

    try:

        if hasattr(
            ev,
            "tokenizer"
        ) and ev.tokenizer is not None:

            ev.tokenizer.save_pretrained(
                model_dir
            )

    except Exception:
        pass

    return model_dir


# ------------------------------------------------------------
# 单个实验
# ------------------------------------------------------------

def run_one(
    method,
    trainable_components,
    encoder_trainable,
    lr,
    seed
):

    global all_results

    print("\n")
    print("=" * 70)
    print("开始实验")
    print("METHOD:", method)
    print("LR:", lr)
    print("SEED:", seed)
    print("=" * 70)

    # --------------------------------------------------------
    # seed
    # --------------------------------------------------------

    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)

    if torch.cuda.is_available():

        torch.cuda.manual_seed(seed)
        torch.cuda.manual_seed_all(seed)

        torch.cuda.empty_cache()

        torch.cuda.reset_peak_memory_stats()

    # --------------------------------------------------------
    # evaluator
    # --------------------------------------------------------

    ev = GLUEvaluator(
        "rte",
        "bert-base-cased",
        "cuda"
    )

    # --------------------------------------------------------
    # dataset
    # --------------------------------------------------------

    ev.preprocess_dataset(
        padding="max_length",
        max_sequence_len=128,
        batch_size=8
    )

    # --------------------------------------------------------
    # training preparation
    # --------------------------------------------------------

    ev.training_preparation(
        learning_rate=lr,
        optimizer="adamw",
        encoder_trainable=encoder_trainable,
        trainable_components=trainable_components,
        verbose=False
    )

    ev.model = ev.model.to(
        ev.device
    )

    # --------------------------------------------------------
    # trainable parameters
    # --------------------------------------------------------

    trainable_items = [
        (name, p)
        for name, p in ev.model.named_parameters()
        if p.requires_grad
    ]

    trainable_params = sum(
        p.numel()
        for _, p in trainable_items
    )

    trainable_tensors = len(
        trainable_items
    )

    print("\nTrainable parameters:")
    print(
        trainable_params
    )

    print(
        "Trainable tensors:",
        trainable_tensors
    )

    for name, p in trainable_items:

        print(
            " ",
            name,
            tuple(p.shape)
        )

    # --------------------------------------------------------
    # 参数快照
    # --------------------------------------------------------

    before_params = (
        snapshot_trainable_parameters(
            ev.model
        )
    )

    # --------------------------------------------------------
    # GPU memory reset
    # --------------------------------------------------------

    if torch.cuda.is_available():

        torch.cuda.empty_cache()
        torch.cuda.reset_peak_memory_stats()

    # --------------------------------------------------------
    # training
    # --------------------------------------------------------

    start_time = time.time()

    ev.train_and_evaluate(
        num_epochs=4,
        evaluation_frequency=1
    )

    elapsed = (
        time.time()
        - start_time
    )

    # --------------------------------------------------------
    # evaluation
    # --------------------------------------------------------

    val_accs = list(
        ev.evaluations[
            "validation"
        ][
            "Accuracy"
        ]
    )

    train_accs = list(
        ev.evaluations.get(
            "train",
            {}
        ).get(
            "Accuracy",
            []
        )
    )

    # --------------------------------------------------------
    # gradient check
    # --------------------------------------------------------

    (
        grad_with,
        grad_without,
        max_grad
    ) = check_gradients(
        ev.model
    )

    # --------------------------------------------------------
    # update check
    # --------------------------------------------------------

    (
        updated,
        not_updated,
        max_change
    ) = check_parameter_updates(
        ev.model,
        before_params
    )

    # --------------------------------------------------------
    # optimizer steps
    # --------------------------------------------------------

    optimizer_steps = (
        get_optimizer_steps(
            ev.optimizer
        )
    )

    # --------------------------------------------------------
    # GPU memory
    # --------------------------------------------------------

    if torch.cuda.is_available():

        peak_memory = (
            torch.cuda
            .max_memory_allocated()
            / 1024**3
        )

    else:

        peak_memory = 0.0

    # --------------------------------------------------------
    # model save
    # --------------------------------------------------------

    model_dir = save_model_to_drive(
        ev,
        method,
        lr,
        seed
    )

    # --------------------------------------------------------
    # result
    # --------------------------------------------------------

    rec = {

        "method": method,

        "lr": float(lr),

        "seed": int(seed),

        "task": "rte",

        "model": "bert-base-cased",

        "epochs": 4,

        "batch_size": 8,

        "max_sequence_len": 128,

        "trainable_params":
            int(trainable_params),

        "trainable_tensors":
            int(trainable_tensors),

        "trainable_names":
            [
                name
                for name, _
                in trainable_items
            ],

        "val_accs":
            [
                float(x)
                for x in val_accs
            ],

        "train_accs":
            [
                float(x)
                for x in train_accs
            ],

        "final_val_acc":
            float(val_accs[-1]),

        "best_val_acc":
            float(max(val_accs)),

        "elapsed":
            float(elapsed),

        "peak_mem":
            float(peak_memory),

        "optimizer_steps":
            optimizer_steps,

        "gradient_tensors":
            len(grad_with),

        "gradient_missing_tensors":
            len(grad_without),

        "max_gradient":
            float(max_grad),

        "updated_tensors":
            len(updated),

        "not_updated_tensors":
            len(not_updated),

        "updated_names":
            updated,

        "not_updated_names":
            not_updated,

        "max_parameter_change":
            float(max_change),

        "model_dir":
            model_dir
    }

    # --------------------------------------------------------
    # 打印验证信息
    # --------------------------------------------------------

    print("\n")
    print("=" * 70)
    print("实验完成")
    print("=" * 70)

    print(
        "METHOD:",
        method
    )

    print(
        "LR:",
        lr
    )

    print(
        "SEED:",
        seed
    )

    print(
        "Trainable parameters:",
        trainable_params
    )

    print(
        "Trainable tensors:",
        trainable_tensors
    )

    print(
        "Gradient tensors:",
        len(grad_with)
    )

    print(
        "Updated tensors:",
        len(updated)
    )

    print(
        "Not updated tensors:",
        len(not_updated)
    )

    print(
        "Optimizer steps:",
        optimizer_steps
    )

    print(
        "Best validation:",
        max(val_accs)
    )

    print(
        "Final validation:",
        val_accs[-1]
    )

    print(
        "Elapsed:",
        f"{elapsed:.2f}s"
    )

    print(
        "Peak GPU memory:",
        f"{peak_memory:.3f} GB"
    )

    print(
        "Model saved:",
        model_dir
    )

    # --------------------------------------------------------
    # 追加结果并立即保存
    # --------------------------------------------------------

    all_results.append(
        rec
    )

    save_results()

    print(
        "JSON 已立即保存：",
        RESULTS_PATH
    )

    # --------------------------------------------------------
    # 清理
    # --------------------------------------------------------

    del ev

    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    return rec
```

```python
# ============================================================
# Cell 5
# 自动续跑
# ============================================================

def run_grid(
    method,
    components,
    encoder_trainable,
    lr,
    seeds
):

    print("\n")
    print("=" * 70)
    print("实验组：", method)
    print("LR：", lr)
    print("Seeds：", seeds)
    print("=" * 70)

    for seed in seeds:

        if has_completed(
            method,
            lr,
            seed
        ):

            print(
                f"\n跳过已完成："
                f"{method} | LR={lr} | seed={seed}"
            )

            continue

        try:

            run_one(
                method=method,
                trainable_components=components,
                encoder_trainable=encoder_trainable,
                lr=lr,
                seed=seed
            )

        except Exception as e:

            print("\n")
            print("=" * 70)
            print("本次实验失败")
            print("=" * 70)

            print(
                "METHOD:",
                method
            )

            print(
                "LR:",
                lr
            )

            print(
                "SEED:",
                seed
            )

            print(
                "ERROR:",
                repr(e)
            )

            traceback.print_exc()

            # 清理 GPU
            if torch.cuda.is_available():
                torch.cuda.empty_cache()

            # ------------------------------------------------
            # 注意：
            # 失败实验不写入 all_results
            # 因此下次重新运行时会自动重试
            # ------------------------------------------------

            continue

    print("\n")
    print("=" * 70)
    print("实验组结束：", method)
    print("=" * 70)
```

```python
# ============================================================
# Cell 6
# 正式运行 12 个实验
# ============================================================

SEEDS = [
    42,
    123,
    2026
]

# ============================================================
# 1. BitFit
# ============================================================

run_grid(
    method="BitFit",

    components=[
        "bias"
    ],

    encoder_trainable=False,

    lr=5e-4,

    seeds=SEEDS
)


# ============================================================
# 2. Classifier-only
# ============================================================

run_grid(
    method="Classifier-only",

    components=[],

    encoder_trainable=False,

    lr=1e-3,

    seeds=SEEDS
)


# ============================================================
# 3. Full Fine-tuning
# ============================================================

run_grid(
    method="Full FT",

    components=None,

    encoder_trainable=True,

    lr=2e-5,

    seeds=SEEDS
)


# ============================================================
# 4. Partial BitFit
# ============================================================

run_grid(
    method="Partial BitFit",

    components=[
        "attention.self.query.bias",
        "intermediate.dense.bias"
    ],

    encoder_trainable=False,

    lr=5e-3,

    seeds=SEEDS
)


print("\n")
print("=" * 70)
print("12 个实验全部执行完毕")
print("=" * 70)

print(
    "当前完成数量：",
    len(all_results)
)

print(
    "结果文件：",
    RESULTS_PATH
)
```

```python
# ============================================================
# Cell 7
# 最终结果汇总
# ============================================================

import pandas as pd
import numpy as np
import os
import json

# ------------------------------------------------------------
# 重新读取最新 JSON
# ------------------------------------------------------------

with open(
    RESULTS_PATH,
    "r",
    encoding="utf-8"
) as f:

    final_results = json.load(f)

# ------------------------------------------------------------
# 检查数量
# ------------------------------------------------------------

print(
    "完成实验数量：",
    len(final_results)
)

expected = 12

if len(final_results) < expected:

    print(
        f"\n注意：目前只有 {len(final_results)}/{expected} 个实验完成。"
    )

else:

    print(
        "\n12/12 个实验全部完成。"
    )

# ------------------------------------------------------------
# DataFrame
# ------------------------------------------------------------

rows = []

for r in final_results:

    rows.append({

        "method":
            r["method"],

        "lr":
            r["lr"],

        "seed":
            r["seed"],

        "trainable_params":
            r["trainable_params"],

        "trainable_tensors":
            r["trainable_tensors"],

        "final_val_acc":
            r["final_val_acc"],

        "best_val_acc":
            r["best_val_acc"],

        "elapsed_s":
            r["elapsed"],

        "peak_gpu_gb":
            r["peak_mem"],

        "optimizer_step":
            (
                r["optimizer_steps"][-1]
                if r["optimizer_steps"]
                else None
            ),

        "gradient_tensors":
            r["gradient_tensors"],

        "updated_tensors":
            r["updated_tensors"],

        "not_updated_tensors":
            r["not_updated_tensors"],

        "max_parameter_change":
            r["max_parameter_change"]
    })


df = pd.DataFrame(rows)

df = df.sort_values(
    [
        "method",
        "seed"
    ]
)

print("\n")
print("=" * 100)
print("逐实验结果")
print("=" * 100)

print(
    df.to_string(
        index=False
    )
)

# ------------------------------------------------------------
# Summary
# ------------------------------------------------------------

summary = (
    df
    .groupby(
        [
            "method",
            "lr"
        ]
    )
    .agg(

        n=(
            "seed",
            "count"
        ),

        final_val_mean=(
            "final_val_acc",
            "mean"
        ),

        final_val_std=(
            "final_val_acc",
            "std"
        ),

        best_val_mean=(
            "best_val_acc",
            "mean"
        ),

        best_val_std=(
            "best_val_acc",
            "std"
        ),

        trainable_params=(
            "trainable_params",
            "first"
        ),

        trainable_tensors=(
            "trainable_tensors",
            "first"
        ),

        time_mean_s=(
            "elapsed_s",
            "mean"
        ),

        memory_mean_gb=(
            "peak_gpu_gb",
            "mean"
        ),

        updated_tensors=(
            "updated_tensors",
            "first"
        )
    )
    .reset_index()
)

summary = summary.sort_values(
    "final_val_mean",
    ascending=False
)

print("\n")
print("=" * 100)
print("最终方法汇总")
print("=" * 100)

print(
    summary.to_string(
        index=False
    )
)

# ------------------------------------------------------------
# 保存 CSV
# ------------------------------------------------------------

ALL_CSV = os.path.join(
    RESULT_ROOT,
    "bitfit_all_results.csv"
)

SUMMARY_CSV = os.path.join(
    RESULT_ROOT,
    "bitfit_summary.csv"
)

df.to_csv(
    ALL_CSV,
    index=False,
    encoding="utf-8-sig"
)

summary.to_csv(
    SUMMARY_CSV,
    index=False,
    encoding="utf-8-sig"
)

print("\n")
print("已保存：")
print(RESULTS_PATH)
print(ALL_CSV)
print(SUMMARY_CSV)
```

```python
# ============================================================
# BitFit 实际 Gradient 检查
# 修正版：使用 GLUEvaluator.data_loaders["train"]
# 只做一次 forward + backward，不进行正式训练
# ============================================================

import sys
import random
import numpy as np
import torch

BITFIT_PATH = "/content/BitFit"

if BITFIT_PATH not in sys.path:
    sys.path.insert(0, BITFIT_PATH)

import glue_evaluator

GLUEvaluator = glue_evaluator.GLUEvaluator


def gradient_check(
    method,
    trainable_components,
    encoder_trainable,
    lr,
    seed=42
):

    print("\n" + "=" * 75)
    print(f"Gradient Check: {method}")
    print(f"LR: {lr}")
    print(f"SEED: {seed}")
    print("=" * 75)

    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    random.seed(seed)
    np.random.seed(seed)

    ev = GLUEvaluator(
        "rte",
        "bert-base-cased",
        "cuda"
    )

    ev.preprocess_dataset(
        padding="max_length",
        max_sequence_len=128,
        batch_size=8
    )

    ev.training_preparation(
        learning_rate=lr,
        optimizer="adamw",
        encoder_trainable=encoder_trainable,
        trainable_components=trainable_components,
        verbose=False
    )

    ev.model = ev.model.to(ev.device)

    ev.model.train()

    trainable_params = [
        (name, p)
        for name, p in ev.model.named_parameters()
        if p.requires_grad
    ]

    total_trainable = sum(
        p.numel()
        for _, p in trainable_params
    )

    print(
        f"Trainable tensors: {len(trainable_params)}"
    )

    print(
        f"Trainable parameters: {total_trainable:,}"
    )

    # --------------------------------------------------------
    # 使用 GLUEvaluator 实际的数据加载器
    # --------------------------------------------------------

    train_loader = ev.data_loaders["train"]

    batch = next(iter(train_loader))

    batch = tuple(
        x.to(ev.device)
        for x in batch
    )

    if "roberta" in ev.model_name:

        input_ids, attention_mask, labels = batch

        token_type_ids = None

    else:

        input_ids, attention_mask, token_type_ids, labels = batch

    # --------------------------------------------------------
    # forward
    # --------------------------------------------------------

    ev.optimizer.zero_grad(set_to_none=True)

    outputs = ev.model(
        input_ids=input_ids,
        attention_mask=attention_mask,
        token_type_ids=token_type_ids
    )

    logits = outputs.logits

    labels = labels.view(-1)

    logits = logits.view(
        -1,
        ev.num_labels
    )

    criterion = torch.nn.CrossEntropyLoss()

    loss = criterion(
        logits,
        labels
    )

    print(
        f"Loss before backward: {loss.item():.6f}"
    )

    # --------------------------------------------------------
    # backward
    # --------------------------------------------------------

    loss.backward()

    # --------------------------------------------------------
    # 检查每一个 trainable tensor 是否存在 gradient
    # --------------------------------------------------------

    gradient_params = []
    no_gradient_params = []

    for name, param in trainable_params:

        if param.grad is not None:

            grad_norm = (
                param.grad.detach()
                .norm()
                .item()
            )

            gradient_params.append(
                (
                    name,
                    param.numel(),
                    grad_norm
                )
            )

        else:

            no_gradient_params.append(
                (
                    name,
                    param.numel()
                )
            )

    # --------------------------------------------------------
    # 输出结果
    # --------------------------------------------------------

    print()
    print("=" * 75)
    print("Gradient 检查结果")
    print("=" * 75)

    print(
        f"Gradient tensors: "
        f"{len(gradient_params)} / "
        f"{len(trainable_params)}"
    )

    print()

    if gradient_params:

        print("前 20 个 gradient 参数:")

        for name, numel, grad_norm in gradient_params[:20]:

            print(
                f"{name:<65} "
                f"numel={numel:<6} "
                f"grad_norm={grad_norm:.6e}"
            )

    if len(gradient_params) > 20:

        print(
            f"... 其余 "
            f"{len(gradient_params) - 20} 个省略"
        )

    if no_gradient_params:

        print()
        print("没有 gradient 的参数:")

        for name, numel in no_gradient_params:

            print(
                f"{name:<65} "
                f"numel={numel}"
            )

    print()
    print("=" * 75)

    if len(gradient_params) == len(trainable_params):

        print("CHECK: PASS")

        print(
            "所有可训练参数都在 backward 后获得了 gradient。"
        )

    else:

        print("CHECK: NEED REVIEW")

        print(
            f"有 {len(no_gradient_params)} "
            f"个可训练参数没有 gradient。"
        )

    print("=" * 75)

    ev.optimizer.zero_grad(set_to_none=True)

    del ev
    del batch
    del outputs
    del logits
    del loss

    torch.cuda.empty_cache()


# ============================================================
# 1. BitFit
# ============================================================

gradient_check(
    method="BitFit",
    trainable_components=["bias"],
    encoder_trainable=False,
    lr=5e-4,
    seed=42
)


# ============================================================
# 2. Classifier-only
# ============================================================

gradient_check(
    method="Classifier-only",
    trainable_components=[],
    encoder_trainable=False,
    lr=1e-3,
    seed=42
)


# ============================================================
# 3. Full FT
# ============================================================

gradient_check(
    method="Full FT",
    trainable_components=None,
    encoder_trainable=True,
    lr=2e-5,
    seed=42
)


# ============================================================
# 4. Partial BitFit
# ============================================================

gradient_check(
    method="Partial BitFit",
    trainable_components=[
        "attention.self.query.bias",
        "intermediate.dense.bias"
    ],
    encoder_trainable=False,
    lr=5e-3,
    seed=42
)


print("\n" + "=" * 75)
print("Gradient 检查全部完成")
print("=" * 75)
```
